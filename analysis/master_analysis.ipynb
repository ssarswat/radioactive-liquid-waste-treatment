# Radioactive Liquid Waste Treatment — Master Analysis\n\nReproducible analysis of the experimental datasets reconstructed from the 2016 M.Tech dissertation.\n

```python
from pathlib import Path\nimport pandas as pd\nimport matplotlib.pyplot as plt\n\nROOT = Path.cwd()\nif not (ROOT / 'data').exists():\n    ROOT = ROOT.parent\n\nmaterial = pd.read_csv(ROOT / 'data/processed/dataset_01_material_comparison.csv')\nparallel = pd.read_csv(ROOT / 'data/processed/dataset_06_parallel_columns.csv')\nleach = pd.read_csv(ROOT / 'data/processed/dataset_08_leach_rate.csv')\nmaterial.head()
```

## 1. Batch material comparison\n\nCompare Kd and remaining activity across the four materials and four analytes.

```python
ranking = (material.sort_values(['analyte','kd_ml_g'], ascending=[True,False])\n           [['analyte','material','kd_ml_g','remaining_activity_pct']])\nranking
```

```python
material.pivot(index='material', columns='analyte', values='kd_ml_g').plot(kind='bar', figsize=(10,5))\nplt.ylabel('Kd (mL/g)')\nplt.title('Batch distribution coefficient comparison')\nplt.tight_layout()
```

## 2. Parallel column performance\n\nInspect outlet activity and decontamination factor against cumulative bed volumes.

```python
parallel.plot(x='bed_volumes', y=['gross_beta_4a_bq_ml','gross_beta_13x_bq_ml'], figsize=(10,5))\nplt.axhline(3.7, linestyle='--', linewidth=1)\nplt.ylabel('Gross beta activity (Bq/mL)')\nplt.xlabel('Cumulative bed volumes')\nplt.tight_layout()
```

## 3. Spent-media management\n\nPlot the cement waste-product leach-rate history.

```python
leach.plot(x='days', y='leach_rate_g_cm2_day', marker='o', figsize=(10,5))\nplt.ylabel('Leach rate (g/cm² day)')\nplt.xlabel('Days')\nplt.tight_layout()
```
